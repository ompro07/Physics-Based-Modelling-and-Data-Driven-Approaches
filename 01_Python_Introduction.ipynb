# Basic Syntax <a class="anchor" id="t1"></a>
The basic rules for writing simple statements and expressions in Python are
* There must be no **spaces or tab** characters at the beginning of a statement: indentation plays a special role in Python and can cause problems when compiling or running code. For now, just make sure all statements start at the **beginning of the line**.
* **Variable names** in Python can be of any length and can consist of upper- and lowercase letters ( A-Z , a-z ), digits ( 0-9 ), and the <br> underscore character ( _ ). An additional restriction is that a variable name may contain digits, but the first character of a variable name may not be a digit.
* The **'#'** character indicates that the rest of the line is a **comment**.
* Python has extensive built-in help for any questions not covered by these modules. You can run **help()** for an overview, or help(x) for any library, object, or type x to get more information.

# Functional Programming <a class="anchor" id="t5"></a>
Besides object-oriented programming, Python also supports functional programming, which is used more often in engineering applications. Functions with input parameters (arguments) can be defined once using `def` or `lambda`, called by their name, and given a specific value for their parameters. Using `return`, you can specify what the function should output. The output value can then be used further, e.g. by assigning it to a variable or by displaying it with `print`.
<br> It is essential to make sure that everything belonging to the function definition is indented. Otherwise, it will be compiled as if it no longer belongs to the function.

## def <a class="anchor" id="t51"></a>
With `def` a function can be *defined* using the following syntax:

````python
def run_two_times(function, argument):        # def Name (Argumente mit KOMMA getrennt) : | Umlaute sind kein Problem
    return function(function(argument))             # was soll die Funktion ausgeben? | einrücken ist essenziell

def add_five(x):                              # einfache Funktion, die 5 zu seinem Argument x addiert
    return (x+5)                                    # was soll die Funktion ausgeben? | einrücken ist essenziell

print(add_five(3))
print(run_two_times(add_five, 10))        # fünf_addieren ist ein Argument von zwei_mal_ausführen
````

### If-else <a class="anchor" id="t612"></a>
If the **If** condition is True, execute everything directly under If, otherwise execute the statement under **else**.

```python
if some_condition:
    code block
else:
    code block```

````python
x = 12
if 10 < x < 11:
    print("hello")
else:
    print("world")
````

### Else if <a class="anchor" id="613"></a>
Compared to **If-else**, **elseif** allows you to introduce an additional **If** check.

```python
if some_condition:  
    code block
elif some_condition:
    code block
else:
    code block```

````python
x = 10
y = 12
if x > y:
    print( "x>y")
elif x < y:
    print( "x<y")
    if x==10:
        print ("x=10")
    else:
        print ("invalid")
else:
    print ("x=y")
````

# Loops <a class="anchor" id="t62"></a>


### For <a class="anchor" id="t621"></a>

```python
for variable in something:
    code block```
    
For each variable in something, the code block is executed. This is called looping. 
When looping over integers, the `range()` function is useful, which creates a range of integers:
* range(n) = 0, 1, ..., n-1
* range(m,n)= m, m+1, ..., n-1
* range(m,n,s)= m, m+s, m+2s, ..., m + ((n-m-1)//s) * s (the last number is the highest number of the form m+xs that is smaller than n)

````python
for i in range(3):
    print(i)
````

### While <a class="anchor" id="t622"></a>
**while** loops remain active as long as their condition is satisfied.

```python
while some_condition:  
    code block```

````python
i = 1
while i < 4:      # Zustand, der True sein muss, damit die Schleife weiterläuft
    print(i)      # Ausgabe
    i = i+1       # Laufindex
print('Die while Schleife ist nicht mehr aktiv')
````

### Break <a class="anchor" id="t623"></a>

````python
for i in range(100):
    print(i)
    if i >= 5:
        break
````

## List <a class="anchor" id="t81"></a>

````python
#Liste definieren
thislist = ["Apfel", "Melone", "Orange", "Banane", "Kirsche"]
print(thislist[0])      # Der erste Eintrag hat den Index 0!

#negativer Index
print(thislist[-1])

#Bereich
print(thislist[2:5])    # Der Index 5 ist nicht in der Menge, Index 2 hingegen schon

#bis..
print(thislist[:4])

#von..
print(thislist[2:])

#Reihenfolge umkehren
print(thislist[-4:-1])

#Eintrag ändern
thislist[1] = "Schwarzer Pfeffer"
print(thislist)

#for-Schleife durch eine Liste hindurch
for x in thislist:
  print(x)

#if-Condition für Liste
if "Apfel" in thislist:
  print("Jawoll, Apfel gibts")

#Länge einer Liste abfragen
print(len(thislist))

#einen Eintrag anhängen
thislist.append("Kumquat")
print(thislist)

#einen Eintrag an einer bestimmten Pos. hinzufügen
thislist.insert(1, "Bestimmte Position")
print(thislist)

#einen Eintrag entfernen
thislist.remove("Bestimmte Position")
print(thislist)

#Eintrag an Pos. x entfernen
thislist.pop(1)
print(thislist)

#Liste kopieren
mylist = thislist.copy()

#Listen zusammenführen
finallist = mylist + thislist
print(finallist)

#Liste leeren, der array hat Bestand
thislist.clear()
print(thislist)

#Liste komplett löschen
del thislist
print(mylist)
````

## Pandas <a class="anchor" id="t86"></a>
Composed of: Python and data analysis. Pandas is a library (`import pandas`) that provides functions and special data structures for manipulating numerical tables and time series. The two basic data structures are:

* Series: a one-dimensional array-like object. It can hold various data types, e.g. integers, floats, strings, Python objects, etc.
A Series can be viewed as a data structure with two arrays: one array acts as the index, i.e. as a label, and one array contains the actual data (values).
* DataFrame: contains an ordered collection of columns. Each column consists of a single data type, but different columns can have different types, e.g. the first column could be of type integer, while the second column is of type boolean, etc. 
A DataFrame has a row index and a column index. It is like a dictionary of Series with a normal index. Further functions at <br>https://www.python-kurs.eu/pandas_DataFrame.php

````python
import pandas as pd
years = range(2014, 2018)
Jahre = pd.Series(["Semester 1-2", "S. 3-4", "S. 5-6", "S. 7-8"], index=years)
print (Jahre)
````

````python
import pandas as pd
cities = {"Bezeichnung": ["London", "Berlin", "Madrid", "Rom"],
          "Einwohner": [8615246, 3562166, 3165235, 2874038],
          "Land": ["England", "Deutschland", "Spanien", "Italien"]}
city_frame = pd.DataFrame(cities)

# Ordinate beschriften
ordinals = ["erste", "zweite", "dritte", "vierte"]


city_frame = pd.DataFrame(cities, index=ordinals)
print(city_frame)

import matplotlib
city_frame.plot(xticks=range(len(city_frame.index)),
                rot=60)
plt.gca().yaxis.set_major_formatter(matplotlib.ticker.FormatStrFormatter('%d')) 
plt.show()
````
